## shape of data 

```python
# Confirm required columns before merging
required_user_columns = ["user_id"]
required_event_columns = ["user_id", "event_name", "event_time"]

for column in required_user_columns:
    if column not in eligible_users.columns:
        raise ValueError(
            f"eligible_users is missing {column}. "
            f"Available columns: {eligible_users.columns.tolist()}"
        )

for column in required_event_columns:
    if column not in events_business.columns:
        raise ValueError(
            f"events_business is missing {column}. "
            f"Available columns: {events_business.columns.tolist()}"
        )

# Rebuild engagement metrics
engagement = (
    events_business.dropna(subset=["user_id"])
    .groupby("user_id")
    .agg(
        total_events=("event_name", "size"),
        distinct_event_types=("event_name", "nunique"),
        last_activity=("event_time", "max")
    )
    .reset_index()
)

# Merge with eligible users
customer_analysis = (
    eligible_users.drop_duplicates(subset=["user_id"])
    .merge(
        engagement,
        on="user_id",
        how="left",
        validate="one_to_one"
    )
)

customer_analysis["total_events"] = (
    customer_analysis["total_events"].fillna(0).astype(int)
)

customer_analysis["distinct_event_types"] = (
    customer_analysis["distinct_event_types"].fillna(0).astype(int)
)

customer_analysis["activated"] = (
    customer_analysis["user_id"].isin(activation_users)
)


display(customer_analysis.head())
print("Customer analysis shape:", customer_analysis.shape)
```

## events 

```python
retention_events = events_business.dropna(
    subset=["user_id", "event_time"]
).copy()

retention_events = retention_events.merge(
    cohort_users[["user_id", "signup_month"]],
    on="user_id",
    how="inner",
    validate="many_to_one"
)

retention_events["activity_month"] = (
    retention_events["event_time"].dt.to_period("M")
)

retention_events["cohort_age"] = (
    retention_events["activity_month"].astype("int64")
    - retention_events["signup_month"].astype("int64")
)

# Only events on or after the signup month
retention_events = retention_events[
    retention_events["cohort_age"] >= 0
].copy()

display(
    retention_events[
        ["user_id", "signup_month", "activity_month", "cohort_age"]
    ].head()
)
```

## group by 

```python
cohort_sizes = (
    cohort_users.groupby("signup_month")["user_id"]
    .nunique()
)

active_users_by_cohort = (
    retention_events.groupby(
        ["signup_month", "cohort_age"]
    )["user_id"]
    .nunique()
    .rename("active_users")
)

retention_counts = active_users_by_cohort.unstack(
    "cohort_age"
)

retention_rates = retention_counts.div(
    cohort_sizes, axis=0
)

retention_rates = retention_rates.sort_index()

display((retention_rates * 100).round(1))
```

## monthly sanpshot

```python
subs_for_snapshot = subscriptions_business.dropna(
    subset=["user_id", "start_date"]
).copy()

# Create month-end dates covering the observed subscription period
first_month = subs_for_snapshot["start_date"].min().to_period("M")
last_observed_date = max(
    subs_for_snapshot["start_date"].max(),
    subs_for_snapshot["end_date"].max()
    if subs_for_snapshot["end_date"].notna().any()
    else subs_for_snapshot["start_date"].max()
)

last_month = last_observed_date.to_period("M")

month_ends = pd.date_range(
    first_month.start_time,
    last_month.end_time.normalize(),
    freq="ME"
)

print("Number of monthly snapshots:", len(month_ends))
```

## snapshots 

```python
snapshot_parts = []

for month_end in month_ends:
    month_start = month_end.to_period("M").start_time

    active = subs_for_snapshot[
        (subs_for_snapshot["start_date"] <= month_end)
        & (
            subs_for_snapshot["end_date"].isna()
            | (subs_for_snapshot["end_date"] >= month_start)
        )
    ].copy()

    # Convert the month-end period to a string
    active["snapshot_month"] = str(
        month_end.to_period("M")
    )

    snapshot_parts.append(
        active[
            ["user_id", "subscription_id", "snapshot_month", "mrr"]
        ]
    )

# Combine monthly snapshots
if snapshot_parts:
    subscription_snapshot = pd.concat(
        snapshot_parts,
        ignore_index=True
    )
else:
    subscription_snapshot = pd.DataFrame(
        columns=[
            "user_id",
            "subscription_id",
            "snapshot_month",
            "mrr"
        ]
    )

print("Subscription snapshot created successfully!")
print("Rows:", len(subscription_snapshot))
print("Columns:", subscription_snapshot.columns.tolist())

display(subscription_snapshot.head(10))
```

## customers 

```python
monthly_customers = (
    subscription_snapshot[
        ["user_id", "snapshot_month"]
    ]
    .drop_duplicates()
)

customer_month_sets = {
    month: set(group["user_id"])
    for month, group in monthly_customers.groupby("snapshot_month")
}

months = sorted(customer_month_sets)
churn_proxy_rows = []

for i in range(1, len(months)):
    previous_month = months[i - 1]
    current_month = months[i]

    previous_customers = customer_month_sets[previous_month]
    current_customers = customer_month_sets[current_month]

    disappeared = previous_customers - current_customers

    churn_proxy_rows.append({
        "previous_month": previous_month,
        "current_month": current_month,
        "starting_customers": len(previous_customers),
        "customers_not_seen_next_month": len(disappeared),
        "disappearance_rate_pct": (
            len(disappeared) / len(previous_customers) * 100
            if previous_customers else np.nan
        )
    })

churn_proxy = pd.DataFrame(churn_proxy_rows)
display(churn_proxy)
```

```python
snapshot_parts = []

for month_end in month_ends:
    month_start = month_end.to_period("M").start_time

    active = subs_for_snapshot[
        (subs_for_snapshot["start_date"] <= month_end)
        & (
            subs_for_snapshot["end_date"].isna()
            | (subs_for_snapshot["end_date"] >= month_start)
        )
    ].copy()

    # Convert the month-end period to a string
    active["snapshot_month"] = str(
        month_end.to_period("M")
    )

    snapshot_parts.append(
        active[
            ["user_id", "subscription_id", "snapshot_month", "mrr"]
        ]
    )

# Combine monthly snapshots
if snapshot_parts:
    subscription_snapshot = pd.concat(
        snapshot_parts,
        ignore_index=True
    )
else:
    subscription_snapshot = pd.DataFrame(
        columns=[
            "user_id",
            "subscription_id",
            "snapshot_month",
            "mrr"
        ]
    )

print("Subscription snapshot created successfully!")
print("Rows:", len(subscription_snapshot))
print("Columns:", subscription_snapshot.columns.tolist())

display(subscription_snapshot.head(10))
```
